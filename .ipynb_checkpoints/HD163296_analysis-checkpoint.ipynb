```python
import numpy as np

# Assume data already loaded and squeezed:
# data = np.squeeze(hdul[0].data)

# Image center
ny, nx = data.shape
yc, xc = ny / 2, nx / 2

# Pixel grid
y, x = np.indices(data.shape)
x_shift = x - xc
y_shift = y - yc

# HD 163296 geometry (literature values)
incl_deg = 46.0      # inclination
pa_deg   = 133.0     # position angle (east of north)

incl = np.deg2rad(incl_deg)
pa   = np.deg2rad(pa_deg)

# Rotate coordinates by PA
x_rot =  x_shift * np.cos(pa) + y_shift * np.sin(pa)
y_rot = -x_shift * np.sin(pa) + y_shift * np.cos(pa)

# Deproject: stretch minor axis
y_deproj = y_rot / np.cos(incl)

# Deprojected radius
r_deproj = np.sqrt(x_rot**2 + y_deproj**2)

r_int = r_deproj.astype(int)
max_r = r_int.max()

# Median radial profile on deprojected radius
radial_profile_deproj = np.array([
    np.median(data[r_int == i]) for i in range(max_r)
])
```

```python
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter1d

smooth_profile_deproj = gaussian_filter1d(radial_profile_deproj, sigma=5)

plt.figure(figsize=(10,5))
plt.plot(smooth_profile_deproj)
plt.xlabel("Deprojected radius (pixels)")
plt.ylabel("Median intensity (smoothed)")
plt.title("Smoothed Deprojected Radial Profile – HD 163296")
plt.show()
```

```python
import numpy as np

# Toy vortex pressure field
def vortex_P(r, theta, r_v=48, theta_v=0, A=1.0, sr=3, st=0.5):
    """
    r_v     = vortex radius (AU)
    theta_v = vortex azimuth (radians)
    A       = amplitude of vortex
    sr      = radial width
    st      = azimuthal width
    """
    return (r**(-2.5)) * (1 + A * np.exp(-((r-r_v)**2)/(2*sr**2) - ((theta-theta_v)**2)/(2*st**2)))
```

```python
import numpy as np

# Planet parameters
r_p = 86.0      # planet radius (AU), near your 86 AU ring
dr  = 15.0      # half-width of the gap (AU)

# Gap edges (candidate attractor radii)
r_inner_edge = r_p - dr
r_outer_edge = r_p + dr
r_inner_edge, r_outer_edge
```

```python
def P_planet_gap(r, r_p=86.0, dr=15.0, A_edge=2.0, n=2.5):
    """
    Background pressure ~ r^-n
    Gap carved by planet at r_p with half-width dr
    Edge bumps at r_p - dr and r_p + dr
    """
    P0 = r**(-n)
    # Gap: reduce pressure in the planet region
    gap = 1.0 - 0.8 * np.exp(-((r - r_p)**2) / (2 * (dr/2)**2))
    # Edge bumps: enhance pressure at gap edges
    bump_inner = 1.0 + A_edge * np.exp(-((r - (r_p - dr))**2) / (2 * (dr/3)**2))
    bump_outer = 1.0 + A_edge * np.exp(-((r - (r_p + dr))**2) / (2 * (dr/3)**2))
    return P0 * gap * bump_inner * bump_outer
```

```python
r_vals = np.linspace(20, 200, 4000)
P_vals = P_planet_gap(r_vals)

# Find local maxima (gap edge attractors)
local_maxima_pg = []
for i in range(1, len(r_vals)-1):
    if P_vals[i] > P_vals[i-1] and P_vals[i] > P_vals[i+1]:
        local_maxima_pg.append(r_vals[i])

local_maxima_pg
```

```python
import numpy as np
import matplotlib.pyplot as plt

# =========================
# 1. PARAMETERS & DATA
# =========================

# Attractor radii from your earlier results
r_snow   = 56.16
r_vortex = 46.40
r_dead   = 8.0
r_zonal  = np.array([84.59, 126.45, 167.37, 207.85, 248.23])
r_planet = np.array([68.84, 101.25])

# Ring radii (HD 163296 example)
ring_r_AU = np.array([10, 48, 86, 145, 230])

# Weights (you can refine these)
w_snow   = 1.0
w_vortex = 1.5
w_zonal  = 1.2
w_dead   = 1.0
w_planet = 1.3

# =========================
# 2. DISTANCE FUNCTIONS
# =========================

def d_snow(r):
    return abs(r - r_snow)

def d_vortex(r):
    return abs(r - r_vortex)

def d_dead(r):
    return abs(r - r_dead)

def d_zonal(r):
    return np.min(abs(r_zonal - r))

def d_planet_edge(r):
    return np.min(abs(r_planet - r))

# =========================
# 3. ATTRACTOR SCORE
# =========================

def attractor_score(r):
    return (
        w_snow   / (1 + d_snow(r)) +
        w_vortex / (1 + d_vortex(r)) +
        w_zonal  / (1 + d_zonal(r)) +
        w_dead   / (1 + d_dead(r)) +
        w_planet / (1 + d_planet_edge(r))
    )

# =========================
# 4. SCORE CURVE & PEAKS
# =========================

# Radial grid
r_vals = np.linspace(5, 250, 2000)
scores = np.array([attractor_score(r) for r in r_vals])

# Find local maxima (attractor peaks)
peaks = []
for i in range(1, len(r_vals) - 1):
    if scores[i] > scores[i - 1] and scores[i] > scores[i + 1]:
        peaks.append(r_vals[i])
peaks = np.array(peaks)

print("Attractor peaks (AU):")
print(peaks)

# =========================
# 5. OVERLAP STRENGTH
# =========================

overlap_info = []

for r_ring in ring_r_AU:
    # closest peak
    diffs = abs(peaks - r_ring)
    idx = np.argmin(diffs)
    r_peak = peaks[idx]
    distance = diffs[idx]
    score_at_ring = attractor_score(r_ring)
    score_at_peak = attractor_score(r_peak)

    overlap_info.append({
        "ring_radius": float(r_ring),
        "closest_peak": float(r_peak),
        "distance": float(distance),
        "score_at_ring": float(score_at_ring),
        "score_at_peak": float(score_at_peak),
    })

print("\nOverlap metrics (ring vs attractor field):")
for info in overlap_info:
    print(
        f"Ring {info['ring_radius']:6.1f} AU -> "
        f"Peak {info['closest_peak']:7.2f} AU | "
        f"Δr = {info['distance']:6.2f} AU | "
        f"Score(ring) = {info['score_at_ring']:.4f}, "
        f"Score(peak) = {info['score_at_peak']:.4f}"
    )

# =========================
# 6. PLOTTING
# =========================

plt.figure(figsize=(10, 6))

# Attractor score curve
plt.plot(r_vals, scores, label="Attractor score", color="black")

# Mark peaks
plt.scatter(peaks, [attractor_score(p) for p in peaks],
            color="orange", s=30, label="Attractor peaks")

# Overlay rings
for r_ring in ring_r_AU:
    plt.axvline(r_ring, color="blue", linestyle="--", alpha=0.5)
    plt.text(r_ring, max(scores)*0.9, f"{int(r_ring)} AU",
             rotation=90, va="bottom", ha="center", color="blue", fontsize=8)

plt.xlabel("Radius (AU)")
plt.ylabel("Attractor score (arb. units)")
plt.title("Disk Attractor Atlas: Score Field, Peaks, and Rings")
plt.legend()
plt.tight_layout()
plt.show()

# =========================
# 7. NOTES FOR REFINING WEIGHTS
# =========================
# To refine weights:
# - Edit w_snow, w_vortex, w_zonal, w_dead, w_planet at the top.
# - Re-run the script.
# - Watch how peaks move and how overlap metrics change.
#
# You can, for example, try:
#   w_vortex = 2.0
#   w_zonal  = 1.5
# if you want vortices and zonal flows to dominate more.
```

```python
def alignment_error(w_snow, w_vortex, w_zonal, w_dead, w_planet):
    def score(r):
        return (
            w_snow   / (1 + d_snow(r)) +
            w_vortex / (1 + d_vortex(r)) +
            w_zonal  / (1 + d_zonal(r)) +
            w_dead   / (1 + d_dead(r)) +
            w_planet / (1 + d_planet_edge(r))
        )

    scores_local = np.array([score(r) for r in r_vals])

    peaks_local = []
    for i in range(1, len(r_vals)-1):
        if scores_local[i] > scores_local[i-1] and scores_local[i] > scores_local[i+1]:
            peaks_local.append(r_vals[i])
    peaks_local = np.array(peaks_local)

    total_error = 0
    for r_ring in ring_r_AU:
        total_error += np.min(abs(peaks_local - r_ring))

    return total_error
```

```python
def analyze_disk(rings, snow, vortex, dead, zonal, planet, r_vals,
                 w_snow, w_vortex, w_zonal, w_dead, w_planet):

    def d_snow(r):        return abs(r - snow)
    def d_vortex(r):      return abs(r - vortex)
    def d_dead(r):        return abs(r - dead)
    def d_zonal(r):       return np.min(abs(zonal - r))
    def d_planet_edge(r): return np.min(abs(planet - r))

    def score(r):
        return (
            w_snow   / (1 + d_snow(r)) +
            w_vortex / (1 + d_vortex(r)) +
            w_zonal  / (1 + d_zonal(r)) +
            w_dead   / (1 + d_dead(r)) +
            w_planet / (1 + d_planet_edge(r))
        )

    scores_local = np.array([score(r) for r in r_vals])

    peaks_local = []
    for i in range(1, len(r_vals)-1):
        if scores_local[i] > scores_local[i-1] and scores_local[i] > scores_local[i+1]:
            peaks_local.append(r_vals[i])
    peaks_local = np.array(peaks_local)

    alignment = []
    for r_ring in rings:
        diffs = abs(peaks_local - r_ring)
        idx = np.argmin(diffs)
        alignment.append((r_ring, peaks_local[idx], diffs[idx]))

    return alignment, peaks_local

```

```python
def disk_likelihood(snow, vortex, dead, zonal, planet, r_vals,
                    w_snow, w_vortex, w_zonal, w_dead, w_planet):

    def d_snow(r):        return abs(r - snow)
    def d_vortex(r):      return abs(r - vortex)
    def d_dead(r):        return abs(r - dead)
    def d_zonal(r):       return np.min(abs(zonal - r))
    def d_planet_edge(r): return np.min(abs(planet - r))

    def score(r):
        return (
            w_snow   / (1 + d_snow(r)) +
            w_vortex / (1 + d_vortex(r)) +
            w_zonal  / (1 + d_zonal(r)) +
            w_dead   / (1 + d_dead(r)) +
            w_planet / (1 + d_planet_edge(r))
        )

    scores_local = np.array([score(r) for r in r_vals])
    likelihood = scores_local / np.max(scores_local)
    return likelihood
```

```python
def attractor_field(snow, vortex, dead, zonal, planet, r_vals,
                    w_snow, w_vortex, w_zonal, w_dead, w_planet):

    def d_snow(r):        return abs(r - snow)
    def d_vortex(r):      return abs(r - vortex)
    def d_dead(r):        return abs(r - dead)
    def d_zonal(r):       return np.min(abs(zonal - r))
    def d_planet_edge(r): return np.min(abs(planet - r))

    def score(r):
        return (
            w_snow   / (1 + d_snow(r)) +
            w_vortex / (1 + d_vortex(r)) +
            w_zonal  / (1 + d_zonal(r)) +
            w_dead   / (1 + d_dead(r)) +
            w_planet / (1 + d_planet_edge(r))
        )

    return np.array([score(r) for r in r_vals])
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# Base radial grid
# ---------------------------------------------------------

r_vals = np.linspace(5, 250, 2000)

# ---------------------------------------------------------
# Attractor field
# ---------------------------------------------------------

def score_field(r_vals, r_snow, r_vortex, r_dead, r_zonal, r_planet,
                w_snow, w_vortex, w_zonal, w_dead, w_planet):

    def d_snow(r):        return abs(r - r_snow)
    def d_vortex(r):      return abs(r - r_vortex)
    def d_dead(r):        return abs(r - r_dead)
    def d_zonal(r):       return np.min(abs(r_zonal - r))
    def d_planet_edge(r): return np.min(abs(r_planet - r))

    def score(r):
        return (
            w_snow   / (1 + d_snow(r)) +
            w_vortex / (1 + d_vortex(r)) +
            w_zonal  / (1 + d_zonal(r)) +
            w_dead   / (1 + d_dead(r)) +
            w_planet / (1 + d_planet_edge(r))
        )

    return np.array([score(r) for r in r_vals])

r_snow   = 56.16
r_vortex = 46.40
r_dead   = 8.0
r_zonal  = np.array([84.59, 126.45, 167.37, 207.85, 248.23])
r_planet = np.array([68.84, 101.25])

w_snow   = 1.0
w_vortex = 1.5
w_zonal  = 1.2
w_dead   = 1.0
w_planet = 1.3

scores = score_field(r_vals, r_snow, r_vortex, r_dead, r_zonal, r_planet,
                     w_snow, w_vortex, w_zonal, w_dead, w_planet)

ring_r_AU = np.array([10, 48, 86, 145, 230])

# ---------------------------------------------------------
# Compression fields
# ---------------------------------------------------------

def alpha_profile(r_vals, alpha_inner, alpha_outer, r_break):
    return alpha_outer + (alpha_inner - alpha_outer) * np.exp(-(r_vals / r_break))

def compression_pgc(r_vals, pressure):
    dp = np.gradient(pressure, r_vals)
    return -dp

def compression_tgc(r_vals, alpha):
    da = np.gradient(alpha, r_vals)
    return -da

def compression_vsc(r_vals, nu):
    dnu = np.gradient(nu, r_vals)
    return -dnu

def total_compression(r_vals, pressure, alpha, nu,
                      w_pgc=1.0, w_tgc=1.0, w_vsc=1.0):

    F_pgc = compression_pgc(r_vals, pressure)
    F_tgc = compression_tgc(r_vals, alpha)
    F_vsc = compression_vsc(r_vals, nu)

    return (
        w_pgc * F_pgc +
        w_tgc * F_tgc +
        w_vsc * F_vsc
    )

alpha = alpha_profile(r_vals, 1e-3, 1e-4, 60)
nu    = alpha_profile(r_vals, 1e-3, 1e-4, 60)

compression = total_compression(r_vals, pressure=scores,
                                alpha=alpha,
                                nu=nu)

# ---------------------------------------------------------
# Stability field
# ---------------------------------------------------------

stability = scores * compression

# ---------------------------------------------------------
# Dust drift velocity
# ---------------------------------------------------------

def dust_drift_velocity(r_vals, pressure):
    dP = np.gradient(pressure, r_vals)
    eta = -dP / (pressure + 1e-12)
    vK = np.sqrt(1/r_vals)
    St = 0.1
    return -2 * eta * vK / St

v_drift = dust_drift_velocity(r_vals, scores)

# ---------------------------------------------------------
# Planet torque field
# ---------------------------------------------------------

def planet_torque(r_vals, planet_r, strength):
    return strength / (1 + abs(r_vals - planet_r)**2)

planet_forces = np.zeros_like(r_vals)
for pr in r_planet:
    planet_forces += planet_torque(r_vals, pr, strength=5.0)

# ---------------------------------------------------------
# Pressure profile
# ---------------------------------------------------------

def pressure_profile(r_vals):
    return 1e-3 * r_vals**(-1.5)

pressure = pressure_profile(r_vals)

# ---------------------------------------------------------
# Surface density profile
# ---------------------------------------------------------

def sigma_profile(r_vals, p=1.0):
    return r_vals**(-p)

sigma = sigma_profile(r_vals)

# ---------------------------------------------------------
# Stability curvature (second derivative)
# ---------------------------------------------------------

curvature = np.gradient(np.gradient(stability, r_vals), r_vals)

# ---------------------------------------------------------
# Ring width predictor
# ---------------------------------------------------------

ring_width = 1 / (np.abs(curvature) + 1e-12)

# ---------------------------------------------------------
# Ring survival probability
# ---------------------------------------------------------

survival = stability - np.abs(v_drift) + planet_forces

# ---------------------------------------------------------
# Plots
# ---------------------------------------------------------

plt.figure(figsize=(12,6))
plt.plot(r_vals, scores, label="Attractor", lw=2)
plt.plot(r_vals, compression, label="Compression", lw=2)
plt.plot(r_vals, stability, label="Stability", lw=2)
plt.legend()
plt.title("Attractor / Compression / Stability")
plt.tight_layout()
plt.show()

plt.figure(figsize=(12,6))
plt.plot(r_vals, v_drift, color="green", lw=2)
plt.title("Dust Drift Velocity")
plt.tight_layout()
plt.show()

plt.figure(figsize=(12,6))
plt.plot(r_vals, planet_forces, color="orange", lw=2)
plt.title("Planet Torque Field")
plt.tight_layout()
plt.show()

plt.figure(figsize=(12,6))
plt.plot(r_vals, pressure, color="red", lw=2)
plt.title("Pressure Profile")
plt.tight_layout()
plt.show()

plt.figure(figsize=(12,6))
plt.plot(r_vals, sigma, color="blue", lw=2)
plt.title("Surface Density Profile")
plt.tight_layout()
plt.show()

plt.figure(figsize=(12,6))
plt.plot(r_vals, curvature, color="purple", lw=2)
plt.title("Stability Curvature")
plt.tight_layout()
plt.show()

plt.figure(figsize=(12,6))
plt.plot(r_vals, ring_width, color="black", lw=2)
plt.title("Ring Width Predictor")
plt.tight_layout()
plt.show()

plt.figure(figsize=(12,6))
plt.plot(r_vals, survival, color="brown", lw=2)
plt.title("Ring Survival Probability")
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# ---------------------------------------------------------
# Base radial grid
# ---------------------------------------------------------

r_vals = np.linspace(5, 250, 2000)

# ---------------------------------------------------------
# Attractor field
# ---------------------------------------------------------

def score_field(r_vals, r_snow, r_vortex, r_dead, r_zonal, r_planet,
                w_snow, w_vortex, w_zonal, w_dead, w_planet):

    def d_snow(r):        return abs(r - r_snow)
    def d_vortex(r):      return abs(r - r_vortex)
    def d_dead(r):        return abs(r - r_dead)
    def d_zonal(r):       return np.min(abs(r_zonal - r))
    def d_planet_edge(r): return np.min(abs(r_planet - r))

    def score(r):
        return (
            w_snow   / (1 + d_snow(r)) +
            w_vortex / (1 + d_vortex(r)) +
            w_zonal  / (1 + d_zonal(r)) +
            w_dead   / (1 + d_dead(r)) +
            w_planet / (1 + d_planet_edge(r))
        )

    return np.array([score(r) for r in r_vals])

r_snow   = 56.16
r_vortex = 46.40
r_dead   = 8.0
r_zonal  = np.array([84.59, 126.45, 167.37, 207.85, 248.23])
r_planet = np.array([68.84, 101.25])

w_snow   = 1.0
w_vortex = 1.5
w_zonal  = 1.2
w_dead   = 1.0
w_planet = 1.3

scores = score_field(r_vals, r_snow, r_vortex, r_dead, r_zonal, r_planet,
                     w_snow, w_vortex, w_zonal, w_dead, w_planet)

ring_r_AU = np.array([10, 48, 86, 145, 230])

# ---------------------------------------------------------
# Compression fields
# ---------------------------------------------------------

def alpha_profile(r_vals, alpha_inner, alpha_outer, r_break):
    return alpha_outer + (alpha_inner - alpha_outer) * np.exp(-(r_vals / r_break))

def compression_pgc(r_vals, pressure):
    dp = np.gradient(pressure, r_vals)
    return -dp

def compression_tgc(r_vals, alpha):
    da = np.gradient(alpha, r_vals)
    return -da

def compression_vsc(r_vals, nu):
    dnu = np.gradient(nu, r_vals)
    return -dnu

def total_compression(r_vals, pressure, alpha, nu,
                      w_pgc=1.0, w_tgc=1.0, w_vsc=1.0):

    F_pgc = compression_pgc(r_vals, pressure)
    F_tgc = compression_tgc(r_vals, alpha)
    F_vsc = compression_vsc(r_vals, nu)

    return (
        w_pgc * F_pgc +
        w_tgc * F_tgc +
        w_vsc * F_vsc
    )

alpha = alpha_profile(r_vals, 1e-3, 1e-4, 60)
nu    = alpha_profile(r_vals, 1e-3, 1e-4, 60)

compression = total_compression(r_vals, pressure=scores,
                                alpha=alpha,
                                nu=nu)

# ---------------------------------------------------------
# Stability field
# ---------------------------------------------------------

stability = scores * compression

# ---------------------------------------------------------
# Dust drift velocity
# ---------------------------------------------------------

def dust_drift_velocity(r_vals, pressure):
    dP = np.gradient(pressure, r_vals)
    eta = -dP / (pressure + 1e-12)
    vK = np.sqrt(1/r_vals)
    St = 0.1
    return -2 * eta * vK / St

v_drift = dust_drift_velocity(r_vals, scores)

# ---------------------------------------------------------
# Planet torque field
# ---------------------------------------------------------

def planet_torque(r_vals, planet_r, strength):
    return strength / (1 + abs(r_vals - planet_r)**2)

planet_forces = np.zeros_like(r_vals)
for pr in r_planet:
    planet_forces += planet_torque(r_vals, pr, strength=5.0)

# ---------------------------------------------------------
# Pressure profile
# ---------------------------------------------------------

def pressure_profile(r_vals):
    return 1e-3 * r_vals**(-1.5)

pressure = pressure_profile(r_vals)

# ---------------------------------------------------------
# Surface density profile
# ---------------------------------------------------------

def sigma_profile(r_vals, p=1.0):
    return r_vals**(-p)

sigma = sigma_profile(r_vals)

# ---------------------------------------------------------
# Stability curvature (second derivative)
# ---------------------------------------------------------

curvature = np.gradient(np.gradient(stability, r_vals), r_vals)

# ---------------------------------------------------------
# Ring width predictor
# ---------------------------------------------------------

ring_width = 1 / (np.abs(curvature) + 1e-12)

# ---------------------------------------------------------
# Ring survival probability
# ---------------------------------------------------------

survival = stability - np.abs(v_drift) + planet_forces

# ---------------------------------------------------------
# Ring classification system
# ---------------------------------------------------------

def classify_ring(r):
    attractor_strength = scores[np.searchsorted(r_vals, r)]
    compression_strength = compression[np.searchsorted(r_vals, r)]
    planet_strength = planet_forces[np.searchsorted(r_vals, r)]
    drift_speed = abs(v_drift[np.searchsorted(r_vals, r)])

    if planet_strength > attractor_strength and planet_strength > compression_strength:
        return "Planet-supported"
    if compression_strength > attractor_strength:
        return "Compression-supported"
    if attractor_strength > compression_strength:
        return "Attractor-supported"
    if drift_speed > stability[np.searchsorted(r_vals, r)]:
        return "Drift-unstable"
    return "Mixed-support"

ring_classes = {r: classify_ring(r) for r in ring_r_AU}

# ---------------------------------------------------------
# Numerical ring stability index
# ---------------------------------------------------------

ring_stability_index = {
    r: stability[np.searchsorted(r_vals, r)] for r in ring_r_AU
}

# ---------------------------------------------------------
# Ring drift prediction table
# ---------------------------------------------------------

ring_drift_table = pd.DataFrame({
    "Radius_AU": ring_r_AU,
    "Drift_Speed": [v_drift[np.searchsorted(r_vals, r)] for r in ring_r_AU],
    "Stability": [stability[np.searchsorted(r_vals, r)] for r in ring_r_AU],
    "Survival": [survival[np.searchsorted(r_vals, r)] for r in ring_r_AU],
    "Classification": [ring_classes[r] for r in ring_r_AU]
})

# ---------------------------------------------------------
# Multi-panel diagnostic figure
# ---------------------------------------------------------

fig, axs = plt.subplots(4, 2, figsize=(18, 20))

axs[0,0].plot(r_vals, scores, lw=2)
axs[0,0].set_title("Attractor Field")

axs[0,1].plot(r_vals, compression, lw=2, color="red")
axs[0,1].set_title("Compression Field")

axs[1,0].plot(r_vals, stability, lw=2, color="purple")
axs[1,0].set_title("Stability Field")

axs[1,1].plot(r_vals, v_drift, lw=2, color="green")
axs[1,1].set_title("Dust Drift Velocity")

axs[2,0].plot(r_vals, planet_forces, lw=2, color="orange")
axs[2,0].set_title("Planet Torque Field")

axs[2,1].plot(r_vals, pressure, lw=2, color="red")
axs[2,1].set_title("Pressure Profile")

axs[3,0].plot(r_vals, curvature, lw=2, color="black")
axs[3,0].set_title("Stability Curvature")

axs[3,1].plot(r_vals, survival, lw=2, color="brown")
axs[3,1].set_title("Ring Survival Probability")

plt.tight_layout()
plt.show()

# ---------------------------------------------------------
# Output ring classification and drift table
# ---------------------------------------------------------

ring_classes, ring_stability_index, ring_drift_table
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# Gap-opening mass estimate (Crida criterion simplified)
# ---------------------------------------------------------

def gap_opening_mass(r_vals, alpha, h_over_r=0.05):
    return (40 * alpha * h_over_r**2) * r_vals

gap_mass = gap_opening_mass(r_vals, alpha)

# ---------------------------------------------------------
# Planet mass inference from ring stability + width
# ---------------------------------------------------------

def infer_planet_mass(stability, ring_width, scale=0.1):
    return scale * stability * ring_width

planet_mass_inferred = infer_planet_mass(stability, ring_width)

# ---------------------------------------------------------
# Ring optical depth modeling
# ---------------------------------------------------------

def optical_depth(sigma, ring_width):
    return sigma * ring_width

tau = optical_depth(sigma, ring_width)

# ---------------------------------------------------------
# Synthetic ALMA radial brightness profile
# ---------------------------------------------------------

def alma_brightness(tau, temperature):
    return tau * temperature

temperature = 150 * r_vals**(-0.5)
brightness = alma_brightness(tau, temperature)

# ---------------------------------------------------------
# Multi-panel figure for new physics layer
# ---------------------------------------------------------

fig, axs = plt.subplots(2, 2, figsize=(16, 12))

axs[0,0].plot(r_vals, gap_mass, lw=2)
axs[0,0].set_title("Gap-Opening Mass Estimate")

axs[0,1].plot(r_vals, planet_mass_inferred, lw=2, color="purple")
axs[0,1].set_title("Planet Mass Inferred from Stability × Width")

axs[1,0].plot(r_vals, tau, lw=2, color="red")
axs[1,0].set_title("Ring Optical Depth")

axs[1,1].plot(r_vals, brightness, lw=2, color="black")
axs[1,1].set_title("Synthetic ALMA Brightness Profile")

plt.tight_layout()
plt.show()

# ---------------------------------------------------------
# Ring-by-ring table for new physics
# ---------------------------------------------------------

ring_gap_mass = [gap_mass[np.searchsorted(r_vals, r)] for r in ring_r_AU]
ring_planet_mass = [planet_mass_inferred[np.searchsorted(r_vals, r)] for r in ring_r_AU]
ring_tau = [tau[np.searchsorted(r_vals, r)] for r in ring_r_AU]
ring_brightness = [brightness[np.searchsorted(r_vals, r)] for r in ring_r_AU]

import pandas as pd

ring_physics_table = pd.DataFrame({
    "Radius_AU": ring_r_AU,
    "GapOpeningMass": ring_gap_mass,
    "InferredPlanetMass": ring_planet_mass,
    "OpticalDepth": ring_tau,
    "Brightness": ring_brightness
})

ring_physics_table
```

```python
# 1. Planet mass inversion (from gap width + torque + drift)

def invert_planet_mass(r_vals, stability, ring_width, planet_forces, v_drift, scale=0.05):
    return scale * (stability * ring_width + planet_forces - np.abs(v_drift))

planet_mass_inversion = invert_planet_mass(r_vals, stability, ring_width, planet_forces, v_drift)

planet_mass_inversion_rings = {
    r: planet_mass_inversion[np.searchsorted(r_vals, r)] for r in ring_r_AU
}
```

```python
# 2. Gap width fitting (match ALMA ring widths)

observed_ring_widths = {
    10: 5,
    48: 8,
    86: 10,
    145: 12,
    230: 15
}

def fit_gap_width(r_vals, ring_width, observed_widths):
    fitted = {}
    for r, w_obs in observed_widths.items():
        w_model = ring_width[np.searchsorted(r_vals, r)]
        fitted[r] = {
            "ObservedWidth": w_obs,
            "ModelWidth": w_model,
            "Residual": w_model - w_obs
        }
    return fitted

gap_width_fit = fit_gap_width(r_vals, ring_width, observed_ring_widths)
```

```python
# 3. Dust size distribution modeling

dust_sizes = np.array([0.01, 0.1, 1.0, 10.0])  # mm

def dust_response(r_vals, stability, v_drift, dust_size):
    St = dust_size
    return stability - St * np.abs(v_drift)

dust_response_field = {
    s: dust_response(r_vals, stability, v_drift, s) for s in dust_sizes
}
```

```python
# 4. Synthetic ALMA 2D images (radial -> 2D disk)

def radial_to_2d(brightness, r_vals, npix=256, r_max=250):
    x = np.linspace(-r_max, r_max, npix)
    y = np.linspace(-r_max, r_max, npix)
    X, Y = np.meshgrid(x, y)
    R = np.sqrt(X**2 + Y**2)
    B2D = np.interp(R, r_vals, brightness, left=0, right=0)
    return X, Y, B2D

X, Y, B2D = radial_to_2d(brightness, r_vals)

plt.figure(figsize=(6,6))
plt.imshow(B2D, extent=[-250,250,-250,250], origin="lower", cmap="inferno")
plt.colorbar(label="Brightness")
plt.title("Synthetic ALMA 2D Disk")
plt.tight_layout()
plt.show()
```

```python
# 5. Temperature structure modeling

def temperature_profile(r_vals, T0=150, q=0.5):
    return T0 * r_vals**(-q)

temperature_struct = temperature_profile(r_vals)

plt.figure(figsize=(10,4))
plt.plot(r_vals, temperature_struct, lw=2)
plt.title("Temperature Structure")
plt.tight_layout()
plt.show()
```

```python
# 6. Gas velocity field

def gas_velocity_field(r_vals, pressure):
    vK = np.sqrt(1/r_vals)
    dP = np.gradient(pressure, r_vals)
    eta = -dP / (pressure + 1e-12)
    v_gas = vK * (1 - eta)
    return vK, v_gas

vK, v_gas = gas_velocity_field(r_vals, pressure)

plt.figure(figsize=(10,4))
plt.plot(r_vals, vK, label="Keplerian", lw=2)
plt.plot(r_vals, v_gas, label="Gas", lw=2)
plt.legend()
plt.title("Gas vs Keplerian Velocity")
plt.tight_layout()
plt.show()
```

```python
# 7. Ring evolution over time (no animation)

times = np.linspace(0, 1e6, 200)  # years

def evolve_ring_radius(r0, v_drift_field, r_vals, times):
    r = r0
    radii = []
    for t in times:
        v = np.interp(r, r_vals, v_drift_field)
        r += v * (times[1] - times[0]) * 1e-3
        radii.append(r)
    return np.array(radii)

ring_evolution = {
    r: evolve_ring_radius(r, v_drift, r_vals, times) for r in ring_r_AU
}
```

```python
# 8. Full ring survival classifier

def survival_class(r):
    idx = np.searchsorted(r_vals, r)
    S = survival[idx]
    D = np.abs(v_drift[idx])
    P = planet_forces[idx]
    C = compression[idx]
    A = scores[idx]

    if S > 0 and P > A and P > C:
        return "Planet-locked"
    if S > 0 and C > A:
        return "Compression-locked"
    if S > 0 and A > C:
        return "Attractor-locked"
    if D > S:
        return "Drift-dominated"
    if S > 0:
        return "Metastable"
    return "Unstable"

ring_survival_classes = {r: survival_class(r) for r in ring_r_AU}
```

```python
# Planet likelihood field

def planet_likelihood_field(r_vals,
                            gap_mass,
                            planet_mass_inversion,
                            planet_forces,
                            survival,
                            ring_width,
                            stability,
                            w_gap=1.0,
                            w_inv=1.0,
                            w_torque=1.0,
                            w_survival=1.0,
                            w_width=0.5,
                            w_stability=0.5):

    L = (
        w_gap      * gap_mass +
        w_inv      * planet_mass_inversion +
        w_torque   * planet_forces +
        w_survival * survival +
        w_width    * ring_width +
        w_stability* stability
    )

    return L

planet_likelihood = planet_likelihood_field(
    r_vals,
    gap_mass,
    planet_mass_inversion,
    planet_forces,
    survival,
    ring_width,
    stability
)
```

```python
# Extract candidate planet zones

def extract_candidates(r_vals, likelihood, threshold_factor=0.6):
    Lmax = np.max(likelihood)
    threshold = threshold_factor * Lmax

    candidates = []
    in_zone = False
    start = None

    for i in range(len(r_vals)):
        if likelihood[i] >= threshold and not in_zone:
            in_zone = True
            start = r_vals[i]
        elif likelihood[i] < threshold and in_zone:
            end = r_vals[i]
            center = 0.5 * (start + end)
            candidates.append((start, end, center))
            in_zone = False

    if in_zone:
        end = r_vals[-1]
        center = 0.5 * (start + end)
        candidates.append((start, end, center))

    return candidates

candidate_zones = extract_candidates(r_vals, planet_likelihood)
```

```python
# Candidate planet table

import pandas as pd

def candidate_planet_table(r_vals,
                           candidate_zones,
                           likelihood,
                           planet_mass_inversion,
                           planet_forces,
                           gap_mass,
                           survival):

    rows = []

    for (start, end, center) in candidate_zones:
        idx = np.searchsorted(r_vals, center)

        rows.append({
            "Start_AU": start,
            "End_AU": end,
            "Center_AU": center,
            "Likelihood": likelihood[idx],
            "InferredPlanetMass": planet_mass_inversion[idx],
            "Torque": planet_forces[idx],
            "GapOpeningMass": gap_mass[idx],
            "Survival": survival[idx]
        })

    return pd.DataFrame(rows)

candidate_planet_table = candidate_planet_table(
    r_vals,
    candidate_zones,
    planet_likelihood,
    planet_mass_inversion,
    planet_forces,
    gap_mass,
    survival
)

candidate_planet_table
```

```python
# Squeeze deficit field (missing physics detector)

def required_squeeze_field(stability_target):
    return stability_target

def available_squeeze_field(stability, compression, planet_forces):
    return stability + compression + planet_forces

S_required = required_squeeze_field(np.abs(stability))
S_available = available_squeeze_field(stability, compression, planet_forces)

squeeze_deficit = S_required - S_available

plt.figure(figsize=(12,5))
plt.plot(r_vals, squeeze_deficit, lw=2, color="red")
plt.title("Squeeze Deficit Field")
plt.tight_layout()
plt.show()

squeeze_deficit_rings = {
    r: squeeze_deficit[np.searchsorted(r_vals, r)] for r in ring_r_AU
}
```

```python
# Chemical nursery atlas (multi-species)

species_snowlines = {
    "H2O": 3,
    "CO2": 10,
    "CO": 30,
    "N2": 50,
    "CH4": 20
}

def species_nursery_field(r_vals, snow_r, pressure, survival, strength=1.0, width=5.0):
    bump = np.exp(-((r_vals - snow_r)**2) / (2*width**2))
    return strength * bump * pressure * (survival + 1)

species_nurseries = {}
for name, r_s in species_snowlines.items():
    species_nurseries[name] = species_nursery_field(r_vals, r_s, pressure, survival)

plt.figure(figsize=(12,6))
for name, field in species_nurseries.items():
    plt.plot(r_vals, field, lw=2, label=name)
plt.legend()
plt.title("Chemical Nursery Fields")
plt.tight_layout()
plt.show()
```

```python
# DiskAtlas class wrapping everything

class DiskAtlas:
    def __init__(self,
                 r_vals,
                 scores,
                 compression,
                 stability,
                 v_drift,
                 planet_forces,
                 pressure,
                 sigma,
                 curvature,
                 ring_width,
                 survival,
                 gap_mass,
                 planet_mass_inversion,
                 brightness,
                 alpha,
                 nu):

        self.r_vals = r_vals
        self.scores = scores
        self.compression = compression
        self.stability = stability
        self.v_drift = v_drift
        self.planet_forces = planet_forces
        self.pressure = pressure
        self.sigma = sigma
        self.curvature = curvature
        self.ring_width = ring_width
        self.survival = survival
        self.gap_mass = gap_mass
        self.planet_mass_inversion = planet_mass_inversion
        self.brightness = brightness
        self.alpha = alpha
        self.nu = nu

        self.M_pebble_iso = pebble_isolation_mass(self.r_vals, self.alpha)
        self.ring_sharpness = ring_sharpness_field(self.curvature, self.v_drift, self.nu)
        self.S_required = required_squeeze_field(np.abs(self.stability))
        self.S_available = available_squeeze_field(self.stability, self.compression, self.planet_forces)
        self.squeeze_deficit = self.S_required - self.S_available

        self.planet_likelihood = planet_likelihood_field(
            self.r_vals,
            self.gap_mass,
            self.planet_mass_inversion,
            self.planet_forces,
            self.survival,
            self.ring_width,
            self.stability
        )

    def candidates(self, threshold_factor=0.6):
        return extract_candidates(self.r_vals, self.planet_likelihood, threshold_factor)

    def candidate_table(self):
        zones = self.candidates()
        return candidate_planet_table(
            self.r_vals,
            zones,
            self.planet_likelihood,
            self.planet_mass_inversion,
            self.planet_forces,
            self.gap_mass,
            self.survival
        )

    def pebble_iso_at(self, r):
        return self.M_pebble_iso[np.searchsorted(self.r_vals, r)]

    def sharpness_at(self, r):
        return self.ring_sharpness[np.searchsorted(self.r_vals, r)]

    def squeeze_deficit_at(self, r):
        return self.squeeze_deficit[np.searchsorted(self.r_vals, r)]

atlas = DiskAtlas(
    r_vals,
    scores,
    compression,
    stability,
    v_drift,
    planet_forces,
    pressure,
    sigma,
    curvature,
    ring_width,
    survival,
    gap_mass,
    planet_mass_inversion,
    brightness,
    alpha,
    nu
)

atlas_candidate_table = atlas.candidate_table()
atlas_candidate_table
```

```python
# Stable climate evolution (bounded, normalized)

def evolve_climate_stable(r_vals,
                          pressure_init,
                          temperature_init,
                          alpha_init,
                          n_steps=300,
                          dt=50):

    P = pressure_init.copy()
    T = temperature_init.copy()
    A = alpha_init.copy()

    history = {"pressure": [], "temperature": [], "alpha": []}

    for _ in range(n_steps):

        dP = np.gradient(P, r_vals)
        dT = np.gradient(T, r_vals)

        dP = np.tanh(dP / (np.max(np.abs(dP)) + 1e-12))
        dT = np.tanh(dT / (np.max(np.abs(dT)) + 1e-12))

        P = P - 0.01 * dP * dt
        T = T - 0.01 * dT * dt + 0.005 * (P / (np.max(P) + 1e-12)) * dt
        A = A + 0.00001 * (P / (np.max(P) + 1e-12)) * dt

        P = np.clip(P, 1e-6, None)
        T = np.clip(T, 1e-6, None)
        A = np.clip(A, 1e-6, None)

        history["pressure"].append(P.copy())
        history["temperature"].append(T.copy())
        history["alpha"].append(A.copy())

    return history

climate_history = evolve_climate_stable(
    r_vals,
    pressure,
    temperature_struct,
    alpha,
    n_steps=200,
    dt=50
)
```

```python
# Stable dust evolution (bounded, normalized)

dust_sizes = np.array([0.01, 0.1, 1.0, 10.0])

def evolve_dust_stable(r_vals,
                       v_drift,
                       stability,
                       n_steps=200,
                       dt=20):

    dust_density = {
        s: np.exp(-((r_vals - 50)**2) / (2*20**2)) for s in dust_sizes
    }

    history = {s: [] for s in dust_sizes}

    v_safe = v_drift / (np.max(np.abs(v_drift)) + 1e-12)

    for _ in range(n_steps):
        for s in dust_sizes:

            v = v_safe * s
            flux = v * dust_density[s]

            adv = np.gradient(flux, r_vals)
            adv = np.tanh(adv / (np.max(np.abs(adv)) + 1e-12))

            trap = 0.001 * np.tanh(stability)

            dust_density[s] += (-adv + trap) * dt
            dust_density[s] = np.clip(dust_density[s], 0, None)

            history[s].append(dust_density[s].copy())

    return dust_density, history

dust_final, dust_history = evolve_dust_stable(
    r_vals,
    v_drift,
    stability,
    n_steps=200,
    dt=20
)
```

```python
# Stable planet growth (bounded, normalized)

def planet_growth_track_stable(r_vals,
                               start_r,
                               M_init,
                               M_pebble_iso,
                               dust_density,
                               v_drift,
                               n_steps=300,
                               dt=20):

    r = start_r
    M = M_init

    track_r = []
    track_M = []

    v_safe = v_drift / (np.max(np.abs(v_drift)) + 1e-12)

    for _ in range(n_steps):

        idx = np.searchsorted(r_vals, r)
        idx = np.clip(idx, 0, len(r_vals)-1)

        M_iso = M_pebble_iso[idx]

        if M < M_iso:
            local_dust = sum(d[idx] for d in dust_density.values())
            dM = 0.0001 * np.tanh(local_dust / (np.max(local_dust) + 1e-12))
            M += dM

        v = np.interp(r, r_vals, v_safe)
        r += v * dt * 0.1

        r = np.clip(r, r_vals[0], r_vals[-1])
        M = np.clip(M, 0, None)

        track_r.append(r)
        track_M.append(M)

    return np.array(track_r), np.array(track_M)

track_r, track_M = planet_growth_track_stable(
    r_vals,
    25.0,
    0.1,
    M_pebble_iso,
    dust_final,
    v_drift,
    n_steps=300,
    dt=20
)
```

```python
# Disk feature extraction and classifier (already stable)

def disk_features(r_vals,
                  scores,
                  compression,
                  stability,
                  planet_forces,
                  survival,
                  squeeze_deficit):

    feat = {}

    feat["mean_stability"] = np.mean(stability)
    feat["max_stability"] = np.max(stability)
    feat["mean_compression"] = np.mean(compression)
    feat["max_torque"] = np.max(planet_forces)
    feat["mean_survival"] = np.mean(survival)
    feat["max_survival"] = np.max(survival)
    feat["squeeze_deficit_mean"] = np.mean(squeeze_deficit)
    feat["squeeze_deficit_min"] = np.min(squeeze_deficit)

    return feat

def classify_disk(feat):
    if feat["max_torque"] > 1.0 and feat["mean_survival"] > 0:
        return "Planet-locked disk"

    if feat["mean_compression"] > feat["mean_stability"]:
        return "Compression-dominated disk"

    if feat["squeeze_deficit_min"] < -0.1:
        return "Missing-physics disk"

    if feat["mean_survival"] < 0:
        return "Drift-dominated disk"

    return "Mixed-type disk"

atlas_features = disk_features(
    r_vals,
    scores,
    compression,
    stability,
    planet_forces,
    survival,
    squeeze_deficit
)

disk_type = classify_disk(atlas_features)
disk_type
```

```python
# Missing-physics map: field + zones

def missing_physics_field(r_vals, squeeze_deficit):
    return squeeze_deficit

missing_physics = missing_physics_field(r_vals, squeeze_deficit)
```

```python
# Extract missing-physics zones (where deficit is strongly negative)

def extract_missing_physics_zones(r_vals,
                                  missing_physics,
                                  threshold_factor=0.5):

    mp_min = np.min(missing_physics)
    threshold = threshold_factor * mp_min  # mp_min is negative

    zones = []
    in_zone = False
    start = None

    for i in range(len(r_vals)):
        if missing_physics[i] <= threshold and not in_zone:
            in_zone = True
            start = r_vals[i]
        elif missing_physics[i] > threshold and in_zone:
            end = r_vals[i]
            center = 0.5 * (start + end)
            zones.append((start, end, center))
            in_zone = False

    if in_zone:
        end = r_vals[-1]
        center = 0.5 * (start + end)
        zones.append((start, end, center))

    return zones

missing_physics_zones = extract_missing_physics_zones(
    r_vals,
    missing_physics,
    threshold_factor=0.5
)
```

```python
# Missing-physics zone table

import pandas as pd

def missing_physics_table(r_vals,
                          missing_physics,
                          zones):

    rows = []

    for (start, end, center) in zones:
        idx = np.searchsorted(r_vals, center)
        idx = np.clip(idx, 0, len(r_vals)-1)

        rows.append({
            "Start_AU": start,
            "End_AU": end,
            "Center_AU": center,
            "Deficit": missing_physics[idx]
        })

    return pd.DataFrame(rows)

missing_physics_df = missing_physics_table(
    r_vals,
    missing_physics,
    missing_physics_zones
)

missing_physics_df
```

```python
class DiskAtlas:
    def __init__(self,
                 r_vals,
                 scores,
                 compression,
                 stability,
                 v_drift,
                 planet_forces,
                 pressure,
                 sigma,
                 curvature,
                 ring_width,
                 survival,
                 gap_mass,
                 planet_mass_inversion,
                 brightness,
                 alpha,
                 nu):

        self.r_vals = r_vals
        self.scores = scores
        self.compression = compression
        self.stability = stability
        self.v_drift = v_drift
        self.planet_forces = planet_forces
        self.pressure = pressure
        self.sigma = sigma
        self.curvature = curvature
        self.ring_width = ring_width
        self.survival = survival
        self.gap_mass = gap_mass
        self.planet_mass_inversion = planet_mass_inversion
        self.brightness = brightness
        self.alpha = alpha
        self.nu = nu

        # Pebble isolation
        self.M_pebble_iso = 20 * alpha * (0.05**3) * r_vals

        # Ring sharpness
        self.ring_sharpness = np.abs(curvature) / (np.abs(v_drift) + nu + 1e-12)

        # Squeeze deficit
        self.S_required = np.abs(stability)
        self.S_available = stability + compression + planet_forces
        self.squeeze_deficit = self.S_required - self.S_available

        # Planet likelihood
        self.planet_likelihood = (
            gap_mass +
            planet_mass_inversion +
            planet_forces +
            survival +
            ring_width +
            stability
        )

    # -------------------------
    # Hidden planet candidates
    # -------------------------
    def extract_candidates(self, threshold_factor=0.6):
        L = self.planet_likelihood
        Lmax = np.max(L)
        threshold = threshold_factor * Lmax

        zones = []
        in_zone = False
        start = None

        for i in range(len(self.r_vals)):
            if L[i] >= threshold and not in_zone:
                in_zone = True
                start = self.r_vals[i]
            elif L[i] < threshold and in_zone:
                end = self.r_vals[i]
                center = 0.5 * (start + end)
                zones.append((start, end, center))
                in_zone = False

        if in_zone:
            end = self.r_vals[-1]
            center = 0.5 * (start + end)
            zones.append((start, end, center))

        return zones

    def candidate_table(self):
        zones = self.extract_candidates()
        rows = []

        for (start, end, center) in zones:
            idx = np.searchsorted(self.r_vals, center)
            idx = np.clip(idx, 0, len(self.r_vals)-1)

            rows.append({
                "Start_AU": start,
                "End_AU": end,
                "Center_AU": center,
                "Likelihood": self.planet_likelihood[idx],
                "InferredPlanetMass": self.planet_mass_inversion[idx],
                "Torque": self.planet_forces[idx],
                "GapOpeningMass": self.gap_mass[idx],
                "Survival": self.survival[idx]
            })

        return pd.DataFrame(rows)

    # -------------------------
    # Missing physics zones
    # -------------------------
    def missing_physics_zones(self, threshold_factor=0.5):
        mp = self.squeeze_deficit
        mp_min = np.min(mp)
        threshold = threshold_factor * mp_min

        zones = []
        in_zone = False
        start = None

        for i in range(len(self.r_vals)):
            if mp[i] <= threshold and not in_zone:
                in_zone = True
                start = self.r_vals[i]
            elif mp[i] > threshold and in_zone:
                end = self.r_vals[i]
                center = 0.5 * (start + end)
                zones.append((start, end, center))
                in_zone = False

        if in_zone:
            end = self.r_vals[-1]
            center = 0.5 * (start + end)
            zones.append((start, end, center))

        return zones

    def missing_physics_table(self):
        zones = self.missing_physics_zones()
        rows = []

        for (start, end, center) in zones:
            idx = np.searchsorted(self.r_vals, center)
            idx = np.clip(idx, 0, len(self.r_vals)-1)

            rows.append({
                "Start_AU": start,
                "End_AU": end,
                "Center_AU": center,
                "Deficit": self.squeeze_deficit[idx]
            })

        return pd.DataFrame(rows)

    # -------------------------
    # Stable climate evolution
    # -------------------------
    def evolve_climate(self, n_steps=200, dt=50):
        P = self.pressure.copy()
        T = self.pressure.copy()
        A = self.alpha.copy()

        for _ in range(n_steps):
            dP = np.gradient(P, self.r_vals)
            dT = np.gradient(T, self.r_vals)

            dP = np.tanh(dP / (np.max(np.abs(dP)) + 1e-12))
            dT = np.tanh(dT / (np.max(np.abs(dT)) + 1e-12))

            P = P - 0.01 * dP * dt
            T = T - 0.01 * dT * dt + 0.005 * (P / (np.max(P) + 1e-12)) * dt
            A = A + 0.00001 * (P / (np.max(P) + 1e-12)) * dt

            P = np.clip(P, 1e-6, None)
            T = np.clip(T, 1e-6, None)
            A = np.clip(A, 1e-6, None)

        return P, T, A

    # -------------------------
    # Stable dust evolution
    # -------------------------
    def evolve_dust(self, n_steps=200, dt=20):
        dust_sizes = np.array([0.01, 0.1, 1.0, 10.0])
        dust_density = {
            s: np.exp(-((self.r_vals - 50)**2) / (2*20**2)) for s in dust_sizes
        }

        v_safe = self.v_drift / (np.max(np.abs(self.v_drift)) + 1e-12)

        for _ in range(n_steps):
            for s in dust_sizes:
                v = v_safe * s
                flux = v * dust_density[s]

                adv = np.gradient(flux, self.r_vals)
                adv = np.tanh(adv / (np.max(np.abs(adv)) + 1e-12))

                trap = 0.001 * np.tanh(self.stability)

                dust_density[s] += (-adv + trap) * dt
                dust_density[s] = np.clip(dust_density[s], 0, None)

        return dust_density

    # -------------------------
    # Stable planet growth
    # -------------------------
    def planet_growth(self, start_r=25.0, M_init=0.1, n_steps=300, dt=20):
        r = start_r
        M = M_init

        v_safe = self.v_drift / (np.max(np.abs(self.v_drift)) + 1e-12)

        dust_density = self.evolve_dust(n_steps=50, dt=10)

        track_r = []
        track_M = []

        for _ in range(n_steps):
            idx = np.searchsorted(self.r_vals, r)
            idx = np.clip(idx, 0, len(self.r_vals)-1)

            M_iso = self.M_pebble_iso[idx]

            if M < M_iso:
                local_dust = sum(d[idx] for d in dust_density.values())
                dM = 0.0001 * np.tanh(local_dust / (np.max(local_dust) + 1e-12))
                M += dM

            v = np.interp(r, self.r_vals, v_safe)
            r += v * dt * 0.1

            r = np.clip(r, self.r_vals[0], self.r_vals[-1])
            M = np.clip(M, 0, None)

            track_r.append(r)
            track_M.append(M)

        return np.array(track_r), np.array(track_M)

    # -------------------------
    # Stable classifier
    # -------------------------
    def classify(self):
        feat = {
            "mean_stability": np.mean(np.nan_to_num(self.stability)),
            "max_stability": np.max(np.nan_to_num(self.stability)),
            "mean_compression": np.mean(np.nan_to_num(self.compression)),
            "max_torque": np.max(np.nan_to_num(self.planet_forces)),
            "mean_survival": np.mean(np.nan_to_num(self.survival)),
            "max_survival": np.max(np.nan_to_num(self.survival)),
            "squeeze_deficit_mean": np.mean(np.nan_to_num(self.squeeze_deficit)),
            "squeeze_deficit_min": np.min(np.nan_to_num(self.squeeze_deficit))
        }

        if feat["max_torque"] > 1.0 and feat["mean_survival"] > 0:
            return "Planet-locked disk"

        if feat["mean_compression"] > feat["mean_stability"]:
            return "Compression-dominated disk"

        if feat["squeeze_deficit_min"] < -0.1:
            return "Missing-physics disk"

        if feat["mean_survival"] < 0:
            return "Drift-dominated disk"

        return "Mixed-type disk"
```
